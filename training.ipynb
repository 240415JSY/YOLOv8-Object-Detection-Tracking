```
import cv2
import numpy as np
import csv
from pathlib import Path
from collections import defaultdict
from ultralytics import YOLO

# 1. 학습된 최종 모델 불러오기
model_path = '/content/runs/final_3class/weights/best.pt'

if not Path(model_path).exists():
    model_path = '/content/final_3class_best.pt'

model = YOLO(model_path)

# 2. 원본 영상 열기
cap = cv2.VideoCapture(video_path)

fps = cap.get(cv2.CAP_PROP_FPS)
width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))

if not cap.isOpened() or fps <= 0:
    raise RuntimeError("영상 파일을 열 수 없거나 FPS가 잘못되었습니다.")

# 3. 5초 분량 설정
max_frames = round(fps * 5)
output_dir = Path('/content/final_video')
output_dir.mkdir(parents=True, exist_ok=True)

temp_video = str(output_dir / 'tracking_temp.mp4')

writer = cv2.VideoWriter(
    temp_video,
    cv2.VideoWriter_fourcc(*'mp4v'),
    fps,
    (width, height)
)

if not writer.isOpened():
    raise RuntimeError("결과 영상 파일을 생성하지 못했습니다.")

# 4. 객체별 이동 궤적 저장
track_history = defaultdict(list)
tracking_data = []
processed = 0

# 5. 프레임별 객체 추적
while processed < max_frames:
    success, frame = cap.read()
    if not success:
        break

    result = model.track(
    frame,
    persist=True,
    tracker='botsort.yaml',
    conf=0.20,
    iou=0.4,
    verbose=False
)[0]

    annotated = result.plot()

    if result.boxes is not None and result.boxes.is_track:
        boxes = result.boxes.xywh.cpu().numpy()
        ids = result.boxes.id.int().cpu().tolist()
        classes = result.boxes.cls.int().cpu().tolist()

        for box, track_id, cls_id in zip(boxes, ids, classes):
            x, y, w, h = box
            x, y = float(x), float(y)

            track_history[track_id].append((x, y))

            # 중심점 좌표 기록
            tracking_data.append([
                processed,
                round(processed / fps, 3),
                track_id,
                model.names[cls_id],
                round(x, 2),
                round(y, 2)
            ])

            # 최근 30프레임 궤적 표시
            points = track_history[track_id][-30:]

            if len(points) >= 2:
                pts = np.array(
                    points, dtype=np.int32
                ).reshape(-1, 1, 2)

                cv2.polylines(
                    annotated,
                    [pts],
                    isClosed=False,
                    color=(0, 255, 255),
                    thickness=3
                )

            cv2.circle(
                annotated,
                (int(x), int(y)),
                4,
                (0, 0, 255),
                -1
            )

    writer.write(annotated)
    processed += 1

cap.release()
writer.release()

# 6. 추적 좌표 CSV 저장
csv_path = output_dir / 'trajectory.csv'

with open(csv_path, 'w', newline='', encoding='utf-8-sig') as f:
    csv_writer = csv.writer(f)
    csv_writer.writerow([
        'frame', 'time_sec', 'track_id',
        'class_name', 'center_x', 'center_y'
    ])
    csv_writer.writerows(tracking_data)

print("추적 완료!")
print("처리 프레임:", processed)
print("결과 길이:", round(processed / fps, 2), "초")
print("영상:", temp_video)
print("궤적 CSV:", csv_path)
```

# 5. Python Usage

YOLO26 was reimagined using Python-first principles for the most seamless Python YOLO experience yet. YOLO26 models can be loaded from a trained checkpoint or created from scratch. Then methods are used to train, val, predict, and export the model. See detailed Python usage examples in the [YOLO26 Python Docs](https://docs.ultralytics.com/usage/python).

```
from ultralytics import YOLO

# Load a model
model = YOLO('yolo26n.yaml')  # build a new model from scratch
model = YOLO('yolo26n.pt')  # load a pretrained model (recommended for training)

# Use the model
results = model.train(data='coco8.yaml', epochs=3)  # train the model
results = model.val()  # evaluate model performance on the validation set
results = model('https://ultralytics.com/images/bus.jpg')  # predict on an image
results = model.export(format='onnx')  # export the model to ONNX format
```
